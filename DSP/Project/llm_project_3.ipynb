```python
from moviepy import *
# import cv2
from deepface import DeepFace
import cv2
```

```python
# Load the video
video_path = "trump_biden.mp4"
video = VideoFileClip(video_path)

# Extract audio
audio_path = "audio.wav"
video.audio.write_audiofile(audio_path)

# Extract video frames
frame_rate = 1  # Extract 1 frame per second
frames = []
for i, frame in enumerate(video.iter_frames(fps=frame_rate)):
    frames.append(frame)
    #cv2.imwrite(f"frame_{i}.jpg", frame)  # Save frames (optional)
```

```python
frame_emotions = []
for i, frame in enumerate(frames):
    frame_path = "frame_images/" + f"frame_{i}.jpg"
    #cv2.imwrite(frame_path, frame)

    try:
        result = DeepFace.analyze(
            img_path=frame_path, 
            actions=["emotion"], 
            detector_backend="opencv"
        )
        frame_emotions.append(result[0]["emotion"])
    except ValueError as e:
        #print(f"No face detected in frame {i}, skipping...")
        frame_emotions.append(None)  # or skip entirely
```

```python
import json
import matplotlib.pyplot as plt

# Save combined emotions to a JSON file
with open("video_emotions.json", "w") as f:
    json.dump(frame_emotions, f, indent=4)
```

## Visualize the Video emotions

```python


# Visualize frame emotions over time
emotion_labels = list(frame_emotions[0].keys())
emotion_values = {label: [] for label in emotion_labels}

for frame in frame_emotions:
    for label in emotion_labels:
        emotion_values[label].append(frame[label])

# Plot emotions over time
plt.figure(figsize=(10, 6))
for label in emotion_labels:
    plt.plot(emotion_values[label], label=label)

plt.xlabel("Frame Number")
plt.ylabel("Emotion Score")
plt.title("Emotion Analysis Over Time")
plt.legend()
plt.show()
```
