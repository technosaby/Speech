# Speaker Diarization

Speaker identification involves distinguishing between different speakers in an audio file. This is also known as voice diarization. 
-  pre-trained models like pyannote.audio or SpeechBrain for this task.

```python



# ToDo:// Interspeech Paper --->
pipeline = Pipeline.from_pretrained("pyannote/speaker-diarization@2.1", 
                                    use_auth_token="")

# Apply the pipeline to the audio file
diarization = pipeline("audio_only.wav")

# dump the diarization output to disk using RTTM format
with open("audio_sd.rttm", "w") as rttm:
    diarization.write_rttm(rttm)

# Print the speaker segments
for turn, _, speaker in diarization.itertracks(yield_label=True):
    print(f"Speaker {speaker} spoke from {turn.start:.1f}s to {turn.end:.1f}s")
```
